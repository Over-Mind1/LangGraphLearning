```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.checkpoint.memory import InMemorySaver
load_dotenv()
#==========setup .env ===========
if os.getenv("GEMINI_API_KEY") is None:
    raise ValueError("GEMINI_API_KEY environment variable not set. Please set it in your .env file.")
else:
    print("GEMINI_API_KEY environment variable is set.")
```

```python

llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite",
                                max_tokens=2048,temperature=0.7,
                                 api_key=os.getenv("GEMINI_API_KEY"))
```

```python
from typing import TypedDict, List, Annotated
from langgraph.graph.message import add_messages

class graph_schema(TypedDict):
    messages: Annotated[List, add_messages]
```

```python
def welcome(schema:graph_schema) -> graph_schema:

    curr_message = schema['messages']

    response = llm.invoke(curr_message).content

    schema['messages'] = f"Your message was {curr_message}. Here's my response:  {response}"

    return schema
```

```python
from langgraph.graph import StateGraph, START, END

graph = StateGraph(graph_schema)

# Adding Nodes
graph.add_node("welcome", welcome)

# Adding Edges
graph.add_edge(START, "welcome")
graph.add_edge("welcome", END)


# Create an in-memory saver
saver = InMemorySaver()
memory_graph= graph.compile(checkpointer=saver)
```

```python
response = memory_graph.invoke(
    {"messages":"my name is mohamed sabry and i'm ai engineer and now i learning about langgraph"},
    {'configurable':{'thread_id': 'over-mind-1234'}},
)

for message in response['messages']:
    message.pretty_print()
```
